```python
import pandas as pd
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = [20,10]
```

```python
df1 = pd.read_csv("Bengaluru_House_Data.csv")
df1.head()
```

```python
df2 = df1.drop(['area_type', 'society', 'balcony', 'availability'], axis = 'columns')
df2.head()
```

```python
df2.isnull().sum()
```

```python
df3 = df2.dropna()
df3.isnull().sum()
```

```python
df3['size'].unique()
```

```python
df3['BHK'] = df3['size'].apply(lambda x: int(x.split(' ')[0]))
df3.head()
```

```python
def is_float(x):
  try:
    float(x)
  except:
    return False
  return True
```

```python
df3[~df3['total_sqft'].apply(is_float)].head()
# this returns roes where the sqrft is in a range and
# isnt a single value [boolean masking]
# ~ means negation
```

```python
def range_to_val(x):
  tokens = x.split('-')
  if len(tokens) == 2:
    return (float(tokens[0]) + float(tokens[1]))/2
  try:
    return float(x)
  except:
    return None
```

```python
df4 = df3.copy()
df4['total_sqft'] = df4['total_sqft'].apply(range_to_val)
df4.head()
```

```python
df4['total_sqft'].unique()
```

```python
df5 = df4.copy()
df5['price_per_sqft'] = df5['price']*100000/df5['total_sqft']
df5.head()
```

```python
df5['location'] = df5['location'].apply(lambda x: x.strip())
location_stats = df5.groupby('location')['location'].agg('count').sort_values(ascending = False)
location_stats
```

```python
df5['location'] = df5.location.apply(
    # If the location occurs 10 or fewer times in the dataset, replace it with 'other'
    # Otherwise, keep the original location name
    lambda x: 'other' if x in location_stats[location_stats <= 10] else x
)

# Display the first 10 rows of the updated DataFrame to verify the transformation
df5.head(10)
```

```python
len(df5['location'].unique())
```

```python
df5[(df5['total_sqft']/df5['BHK']) < 300].head()
# these are outliers ar anomalies
```

```python
# We wrap the full comparison in parentheses so that the negation (~) applies to the final boolean result
df6 = df5[~((df5['total_sqft']/df5['BHK']) < 300)]
df6.shape
```

```python
def remove_outliers(df):
  df_out = pd.DataFrame()
  for key, subdf in df.groupby('location'):

    # key: A string representing the name of the current location
    #  (e.g., 'Uttarahalli').
    # subdf: A temporary DataFrame containing only the rows
    #   belonging to that specific location.

    m = np.mean(subdf.price_per_sqft)
    st = np.std(subdf.price_per_sqft)
    reduced_df = subdf[(subdf.price_per_sqft > (m - st)) & (subdf.price_per_sqft <= (m + st))]

    # Filters out the extreme prices (outliers) for this location. It keeps only the
    # rows where the price per sqft is within 1 standard deviation of the mean
    # (between mean - std_dev and mean + std_dev). This keeps roughly 68% of the data points
    #  for each location, filtering out exceptionally cheap or expensive properties

    df_out = pd.concat([df_out, reduced_df], ignore_index = True)
  return df_out

df7 = remove_outliers(df6)
df7.shape
#
```

```python
df8 = df7.drop(['size', 'price_per_sqft'], axis = 'columns')
df8.head(100)
```

```python
dummies = pd.get_dummies(df8.location)
```

```python
df9 = pd.concat([df8, dummies.drop('other', axis = 'columns')], axis = 'columns')
df9.head()
```

```python
df10 = df9.drop('location', axis = 'columns')
df10.head()
```

```python
X = df10.drop('price', axis = 'columns')
y = df10.price
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 10)
#
```

If you don't set random_state (or set it to None), your data will be shuffled differently every single time you run the code. This means you will get different training/testing sets and slightly different model accuracy scores every run.


If you set random_state = 10 (or any other integer), the random shuffling is locked in. Every time you run this cell, you will get the exact same split of training and testing data.

```python
from sklearn.linear_model import LinearRegression
lr_clf = LinearRegression()
lr_clf.fit(X_train, y_train)
lr_clf.score(X_test, y_test)
```

```python
from sklearn.model_selection import ShuffleSplit
from sklearn.model_selection import cross_val_score

cv = ShuffleSplit(n_splits = 5, test_size = 0.2, random_state = 0)
# n_splits = 5: It will shuffle and split your data 5 separate times
cross_val_score(LinearRegression(), X, y, cv = cv)
#
```

Since all 5 scores are fairly close to each other (averaging around 78.8%), it proves that your Linear Regression model is stable and not overly sensitive to how the data is split.


```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
import pandas as pd

def Find_best_model_using_gridsearchcv(X,y):
  algos = {
      'linear_regression': {
          'model': LinearRegression(),
          'params': {}
      },
       'lasso' : {
          'model': Lasso(),
          'params': {
              'alpha': [1, 2],
              'selection': ['random', 'cyclic']
          }
      },
       'decision_tree' : {
          'model': DecisionTreeRegressor(),
          'params': {
              'criterion': ['squared_error', 'friedman_mse'],
              'splitter': ['best', 'random']
          }
      }
  }
  scores = []
  cv = ShuffleSplit(n_splits = 5, test_size = 0.2, random_state =0)
  for algo_name , config in algos.items():
      gs = GridSearchCV(config['model'], config['params'], cv = cv, return_train_score = False)
      gs.fit(X, y)
      scores.append({
          'model': algo_name,
          'best_score': gs.best_score_,
          'best_params': gs.best_params_
      })

  return pd.DataFrame(scores, columns = ['model', 'best_score', 'best_params'])

Find_best_model_using_gridsearchcv(X, y)
```

```python
def predict_price(location, sqft, bath, bhk):
  # Create a single-row DataFrame with all 0s, matching the column structure of X
  x = pd.DataFrame(0, index=[0], columns=X.columns)

  # Assign our continuous inputs directly using their column names
  x['total_sqft'] = sqft
  x['bath'] = bath
  x['BHK'] = bhk

  # Safely set the location dummy column to 1 if it exists in the features
  if location in x.columns:
    x[location] = 1

  return lr_clf.predict(x)[0]
```

1. pd.DataFrame(0, index=[0], columns=X.columns)
pd.DataFrame(...): This is the constructor used to initialize a new pandas DataFrame structure from scratch.
0: The first argument sets the default value for all cells in this newly created DataFrame to 0.
index=[0]: DataFrames are two-dimensional grids with rows and columns. This creates exactly one row, assigning it the index label 0.
columns=X.columns: This passes the list of all column headers from your training set X (all 244 of them).
Syntax Result: You get a $1 \times 244$$1 \times 244$ DataFrame grid of zeroes, where every single column perfectly matches your model's expected features.


x[location] = 1
If the location is found in the columns, we locate that column by its name dynamically and change its value from 0 to 1 (One-hot encoding).

```python
predict_price('1st Phase JP Nagar', 1000, 2, 2)
```
